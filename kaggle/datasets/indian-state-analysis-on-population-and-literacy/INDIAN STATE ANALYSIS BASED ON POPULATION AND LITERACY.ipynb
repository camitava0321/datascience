# IMPORTING PACKAGES

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# CONVERTING CSV FILE TO DATAFRAME


```python
df=pd.read_csv("Indian_states_analysis.csv") #changing csv file to dataframe
```

```python
df.head(10)
```

# DESCRIPTION OF DATAFRAME

```python
df.describe()
```

```python
df.head(10)
```

# DROP STATE CODE

```python
df=df.drop(["state_code"],axis=1) #drop the attribute state_code
df.head()
```

```python
df=df.drop(["location"],axis=1) #drop the attribute state_code
df.head()
```

# NAME OF THE ATTRIBUTES

```python
df.columns
```

# CHECKING FOR NULL VALUES

```python
df.isna() #checking for null values
```

# MAKING SUB DATAFRAMES

```python
df_pop = df.filter(["name_of_city",'state_name','population_total','population_male','population_female','sex_ratio','child_sex_ratio','location'], axis=1)
```

```python
df_pop.shape #checking the number of rows and columns
```

```python
df_lit = df.filter(['name_of_city','location','state_name','literates_total','literates_male','literates_female','effective_literacy_rate_total','effective_literacy_rate_male','effective_literacy_rate_female','total_graduates','male_graduates','female_graduates'],axis=1)
```

# NUMBER OF CITIES PER STATE(BAR GRAPH)

```python
df1 = df.groupby(["state_name"],as_index=False).count() #grouping the cities with respect to states
```

```python
df1.plot.barh(y='name_of_city',x="state_name",fontsize=15) 
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("NUMBER OF CITIES")
plt.ylabel("STATE NAME")
```

# TOP 10 POPULOUS CITY

```python
df_pop_10=df_pop.sort_values('population_total', ascending=False).head(10)
df_pop_10=df_pop_10.drop(["population_male","population_female","sex_ratio","child_sex_ratio"],axis=1).head(10)
df_pop_10 #printing the cities with top 10 population
```

# POPULATION PER STATE

```python
df_pop_total=df[['state_name','population_total']].groupby(['state_name']).sum() #plotting graph state name vs population
df_pop_total.head()
df_pop_total.plot(kind='barh',fontsize=15)
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("POPULATION TOTAL")
plt.ylabel("STATE NAME")
```

#  COMPARISON OF LITERACY OF MALE AND FEMALE PER STATE

```python
df_lit_plt = df_lit.groupby(["state_name"]).sum() #plotting graph litearcy rates vs states 
df_lit_plt[['literates_male','literates_female']].plot(kind='barh',stacked=True,fontsize=15)
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.ylabel("STATES")
plt.xlabel("LITERACY RATE")
```

#  COMPARISON  OF MALE AND FEMALE GRADUATES PER STATE

```python
df_lit_plt = df_lit.groupby(["state_name"]).sum() #plotting graph states  vs graduates
df_lit_plt[['male_graduates','female_graduates']].plot(kind='barh',stacked=True,fontsize=15)
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.ylabel("STATES")
plt.xlabel("GRADUATES")
```

# MALE GRADUATES VS MALE LITERATES PER STATE

```python
df_lit_plt = df_lit.groupby(["state_name"]).sum() #plotting graph states  vs graduates
df_lit_plt[['male_graduates','literates_male']].plot(kind='bar',stacked=False,fontsize=15, align = "center")
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("STATES")
plt.ylabel("GRADUATES vs LITERATES")
plt.title("MALE GRADUATES VS MALE LITERATES PER STATE")

```

# MALE GRADUATES VS MALE LITERATES PER STATE

```python
df_lit_plt = df_lit.groupby(["state_name"]).sum() #plotting graph states  vs graduates
df_lit_plt[['female_graduates','literates_female']].plot(kind='bar',stacked=False,fontsize=15, align = "center")
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("STATES")
plt.ylabel("GRADUATES vs LITERATES")
plt.title("FEMALE GRADUATES VS FEMALE LITERATES PER STATE")
```

# PIE CHART OF LITERATES PERCENTAGE PER STATE

```python
df_lit_plt.literates_total.sum()
df_lit_plt_state =(df_lit_plt.literates_total/df_lit_plt.literates_total.sum())*100 # calculating literacy percentage
df_lit_plt_state
df_lit_1 = pd.DataFrame(df_lit_plt_state)
df_lit_1
label=df_lit_1.index
sizes=df_lit_1['literates_total']
plt.pie(sizes,labels= label,autopct='%1.1f%%', shadow=True, startangle=140) #pie plot of literacy rate
fig=plt.gcf()
fig.set_size_inches(10,10)
plt.show()
```

# LINE GRAPH SHOWING LITERACY RATE PER STATE

```python
plt.plot(df_lit_plt_state) #plotting line graph
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("States")
plt.ylabel("Literacy Percentage")
plt.show()
```

# SEX RATIO VS POPULATION

```python
df[["population_total","sex_ratio"]].plot(kind="line")
fig=plt.gcf()
fig.set_size_inches(20,10)
plt.xlabel("SEX RATIO")
plt.ylabel("POPULATION")
plt.show()# plotting sex ratio vs population
```
